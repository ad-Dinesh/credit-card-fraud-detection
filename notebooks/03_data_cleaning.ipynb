```python
import pandas as pd
df = pd.read_csv("../data/raw/creditcard.csv")
```

```python
df_clean = df.copy()
```

```python
df_clean.isnull().sum().sum()
```

```python
duplicate_count = df_clean.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
duplicate_rows = df_clean[df_clean.duplicated(keep=False)]

duplicate_rows.head(10)
```

```python
duplicate_rows["Class"].value_counts()
```

```python
duplicate_rows.groupby("Class").size()
```

```python
df_clean = df_clean.drop_duplicates()
```

```python
print("Duplicate rows after cleaning:", df_clean.duplicated().sum())
```

```python
print("Original shape:", df.shape)
print("Cleaned shape:", df_clean.shape)
```

```python
df_clean.dtypes
```

```python
import numpy as np

print("Positive/negative infinity values:", np.isinf(df_clean.select_dtypes(include=np.number)).sum().sum())
```

```python
print("Total missing values:", df_clean.isnull().sum().sum())
```

```python
print("Unique Class values:", sorted(df_clean["Class"].unique()))
```

```python
print("Final cleaned dataset shape:", df_clean.shape)
print("Total missing values:", df_clean.isnull().sum().sum())
```

```python
print("Total duplicate rows:", df_clean.duplicated().sum())
```

```python
print(df_clean["Class"].value_counts())
```

```python
df_clean.to_csv("../data/processed/creditcard_cleaned.csv", index=False)
```
